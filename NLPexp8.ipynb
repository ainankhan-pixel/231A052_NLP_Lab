```python
#Ainan Khan 231A052
documents = [
"Natural language processing helps computers understand human language.",
    "NLP techniques allow machines to extract meaning from text.",
    "Text processing algorithms can analyze unstructured human language.",
    "Transformer models have revolutionized modern language processing.",
    "Large language models generate human-like responses from prompts."
]
```

```python
#Ainan Khan 231A052
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
#Ainan Khan 231A052
for i, document in enumerate (documents):
  print (f"D{i + 1} : {document}")
```

```python
#Ainan Khan 231A052
vectorizer = TfidfVectorizer(stop_words="english")
```

```python
#Ainan Khan 231A052
tfidf_matrix=vectorizer.fit_transform(documents)
```

```python
#Ainan Khan 231A052
tfidf_matrix.shape
```

```python
#Ainan Khan 231A052
print(tfidf_matrix)
```

```python
#Ainan Khan 231A052
tfidf_df=pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index = ["D1","D2","D3","D4","D5"]
)
print(tfidf_df.round(2))
```

```python
#Ainan Khan 231A052
len(vectorizer.get_feature_names_out())
```

```python
#Ainan Khan 231A052
len(tfidf_df.columns)
```

```python
#Ainan Khan 231A052
similarity = cosine_similarity(tfidf_matrix)
```

```python
#Ainan Khan 231A052
similarity
```

```python
#Ainan Khan 231A052similarity_df = pd.DataFrame (
    similarity,
    columns = ["D1","D2","D3","D4","D5"],
    index = ["D1","D2","D3","D4","D5"]
)
similarity_df
```

```python
#Ainan Khan 231A052
highest_similarity = -1
most_similar_pair = None

for i in range(len(documents)):
    for j in range(i + 1, len(documents)): # Corrected range
        current_similarity = similarity[i][j] # Corrected variable name and assignment
        if current_similarity > highest_similarity:
            highest_similarity = current_similarity
            most_similar_pair = (i, j)

print(
    f"Most similar documents:\n"
    f"D{most_similar_pair[0]+1} and D{most_similar_pair[1]+1}"
)
print(f"Similarity: {highest_similarity:.3f}") # Corrected f-string formatting
```
